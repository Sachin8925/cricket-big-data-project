```python
from pyspark.sql.functions import *
                 
```

```python
silver_df=spark.table('workspace.default.cricket_silver_current_matches')
display(silver_df)
```

```python
gold_match_type_df=silver_df.groupBy('match_type').agg(count('*').alias('Total_Matches'))     
display(gold_match_type_df)     
```

```python
gold_venues_df=silver_df.groupBy('venue').agg(count('*').alias('Total_Matches'))
display(gold_venues_df)
```

```python
team_1_df=silver_df.select(col("status").alias("team"))
team_2_df=silver_df.select(col("status").alias("team"))

all_teams_df=team_1_df.union(team_2_df)
display(all_teams_df)
gold_team_df=all_teams_df.groupBy('team').agg(count('*').alias('Total_Matches'))
```

```python
%sql
select 
  count(*) as total_matches, 
  count(distinct match_type) as total_match_type,
  count(distinct venue) as total_venues
from workspace.default.cricket_silver_current_matches
```
